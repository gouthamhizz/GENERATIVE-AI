#ACTIVATION FUNCTION AND DERIVATIVES

CODE THE MATHEMATICAL FORMULAS AND THEIR RESPECTIVE FOR SIGMOID , TANH , ReLU , AND SOFTMAX FUNCTIONS. PLOT THE VANISHING GRADIENT ZONES.

```python
import numpy as np
import matplotlib.pyplot as plt


class ActivationFunctions:

    #SIGMOID
    def sigmoid(self, x):

        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):

        s = self.sigmoid(x)

        return s * (1 - s)

    #TANH
    def tanh(self, x):
        return np.tanh(x)

    def tanh_derivative(self, x):
        return 1 - np.tanh(x)**2

    #RELU
    def relu(self, x):
        return np.maximum(0, x)

    def relu_derivative(self, x):
        return np.where(x > 0, 1, 0)

    # SOFTMAX
    def softmax(self, x):
        x = x - np.max(x)
        exp_x = np.exp(x)
        return exp_x / np.sum(exp_x)


#CREATE OBJECT
af = ActivationFunctions()

# Input range for analysis
x = np.linspace(-10, 10, 1000)

# Function values
sigmoid = af.sigmoid(x)
tanh = af.tanh(x)
relu = af.relu(x)

# Derivatives
sigmoid_grad = af.sigmoid_derivative(x)
tanh_grad = af.tanh_derivative(x)
relu_grad = af.relu_derivative(x)
```

#PLOTTING ACTIVATION FUNCTION AND IT'S DERIVATIVES GRADIENT

```python
plt.figure(figsize=(10,6))

plt.plot(
    x,
    sigmoid_grad,
    label="Sigmoid Derivative"
)

plt.plot(
    x,
    tanh_grad,
    label="Tanh Derivative"
)

plt.plot(
    x,
    relu_grad,
    label="ReLU Derivative"
)

plt.title("Vanishing Gradient Analysis")

plt.xlabel("Input x")

plt.ylabel("Derivative Value")

plt.grid(True)

plt.legend()

plt.show()
```

```python
"""
Vanishing Gradient Visualization
Gradient vs Network Layer
"""

import numpy as np
import matplotlib.pyplot as plt


# Number of layers
layers = np.arange(1, 21)

# Gradient propagation assumptions
# Sigmoid max derivative = 0.25
# Tanh average derivative = 0.5
# ReLU derivative = 1

sigmoid_grad = (0.25) ** layers
tanh_grad = (0.5) ** layers
relu_grad = np.ones_like(layers)

# Plot
plt.figure(figsize=(10,6))

plt.plot(
    layers,
    sigmoid_grad,
    marker='o',
    label='Sigmoid'
)

plt.plot(
    layers,
    tanh_grad,
    marker='s',
    label='Tanh'
)

plt.plot(
    layers,
    relu_grad,
    marker='^',
    label='ReLU'
)

plt.title(
    "Vanishing Gradient Problem"
)

plt.xlabel(
    "Network Layer Depth"
)

plt.ylabel(
    "Gradient Magnitude"
)

plt.grid(True)

plt.legend()

plt.show()
```
