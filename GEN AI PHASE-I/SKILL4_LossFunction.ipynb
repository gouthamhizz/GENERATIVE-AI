#CUSTOM LOSS FUNCTIONS

MATHEMATICALLY IMPLEMENT MEAN SQUARED ERROR (MSE) AND CATEGORICAL CROSS-ENTROPY LOSS.

```python
import numpy as np


class LossFunctions:

    # Mean Squared Error
    def mse(self, y_true, y_pred):

        squared_error = (y_true - y_pred) ** 2
        loss = np.mean(squared_error)
        return loss


loss_obj = LossFunctions()

y_true = np.array([100,120,150])
y_pred = np.array([90,130,140])

mse_loss = loss_obj.mse(
    y_true,
    y_pred
)

print("MSE Loss =", mse_loss)
```

```python
import numpy as np


class LossFunctions:
    def categorical_cross_entropy(
            self,
            y_true,
            y_pred):

        #Avoid log(0)
        epsilon = 1e-10

        y_pred = np.clip(
            y_pred,
            epsilon,
            1 - epsilon
        )

        loss = -np.sum(
            y_true * np.log(y_pred)
        )

        return loss


loss_obj = LossFunctions()

y_true = np.array([0,1,0])
y_pred = np.array([0.1,0.8,0.1])

loss = loss_obj.categorical_cross_entropy(
    y_true,
    y_pred
)

print(loss)
```
