# SKILL 2 : **Forward Propagation**

Implement a 3-layer Multilayer Perceptron (MLP) forward pass from scratch, defining weight matrices and bias vectors mathematically.

```python
import numpy as np

class MLP:

    def __init__(self):

        # Weights for hidden layer 1
        self.W1 = np.array([
            [0.2, 0.5, 0.1, 0.3],
            [0.4, 0.7, 0.2, 0.6],
            [0.3, 0.8, 0.9, 0.5]
        ])

        # Bias for Hidden Layer 1
        self.b1 = np.array([[0.1, 0.1, 0.1, 0.1]])

        # Weights for hidden layer 2
        self.W2 = np.array([
            [0.4, 0.2, 0.5],
            [0.3, 0.7, 0.1],
            [0.8, 0.6, 0.2],
            [0.5, 0.4, 0.9]
        ])

        # Bias for Hidden Layer 2
        self.b2 = np.array([[0.1, 0.1, 0.1]])

        # Weights before output layer
        self.W3 = np.array([
            [0.2, 0.7],
            [0.6, 0.3],
            [0.5, 0.9]
        ])

        # Bias for Output Layer
        self.b3 = np.array([[0.1, 0.1]])


    def relu(self, x):
        return np.maximum(0, x)

    def softmax(self, x):
        # Numerical stability trick
        x = x - np.max(x)
        exp_values = np.exp(x)
        return exp_values / np.sum(exp_values)

    def forward(self, X):

        print("INPUT")
        print(X)

        # HIDDEN LAYER

        Z1 = np.dot(X, self.W1) + self.b1

        print("\nZ1 (Weighted Sum)")
        print(Z1)

        # Activation
        A1 = self.relu(Z1)
        print("\nA1 (ReLU Output)")
        print(A1)

        Z2 = np.dot(A1, self.W2) + self.b2
        print("\nZ2 (Weighted Sum)")
        print(Z2)

        A2 = self.relu(Z2)
        print("\nA2 (ReLU Output)")
        print(A2)

        Z3 = np.dot(A2, self.W3) + self.b3
        print("\nZ3 (Logits)")
        print(Z3)

        probabilities = self.softmax(Z3)
        print("\nFinal Probabilities")
        print(probabilities)
        return probabilities

X = np.array([[2, 3, 1]])
mlp = MLP()
output = mlp.forward(X)
```
