#THE CALCULUS OF BACKPROPAGATION

IMPLEMENT THE CHAIN RULE FROM SCRATCH TO CALCULATE GRADIENTS FOR THE WEIGHTS OF THE HIDDEN LAYERS IN THE MLP

```python
import numpy as np


class NeuralNetworkInvestigation:

    def __init__(self):

        np.random.seed(42)

        # NETWORK ARCHITECTURE
        # Input Layer    = 3
        # Hidden Layer 1 = 4
        # Hidden Layer 2 = 3
        # Output Layer   = 2

        self.W1 = np.random.randn(3, 4)
        self.b1 = np.zeros((1, 4))

        self.W2 = np.random.randn(4, 3)
        self.b2 = np.zeros((1, 3))

        self.W3 = np.random.randn(3, 2)
        self.b3 = np.zeros((1, 2))

    #ReLU(x)
    def relu(self, x):
        return np.maximum(0, x)

    # RELU DERIVATIVE
    def relu_derivative(self, x):
        return np.where(x > 0, 1, 0)

    # SOFTMAX (Converts logits to probabilities)
    def softmax(self, x):
        x = x - np.max(x)
        exp_values = np.exp(x)
        return exp_values / np.sum(exp_values)

    # FORWARD PROPAGATION
    def forward(self, X):

        self.X = X

        print("\n" + "=" * 60)
        print("INPUT FEATURE REPORT")
        print("=" * 60)

        feature_names = [
            "Technical Skill",
            "Communication",
            "Experience"
        ]

        for feature, value in zip(feature_names, X[0]):
            print(f"{feature:<20}: {value}")

        # HIDDEN LAYER 1 (Z1 = XW1 + b1)
        self.Z1 = np.dot(X, self.W1) + self.b1
        self.A1 = self.relu(self.Z1)

        # HIDDEN LAYER 2
        self.Z2 = np.dot(self.A1, self.W2) + self.b2
        self.A2 = self.relu(self.Z2)


        # OUTPUT LAYER
        self.Z3 = np.dot(self.A2, self.W3) + self.b3
        self.y_hat = self.softmax(self.Z3)

        print("\nZ1")
        print(self.Z1)

        print("\nA1")
        print(self.A1)

        print("\nZ2")
        print(self.Z2)

        print("\nA2")
        print(self.A2)

        print("\nZ3")
        print(self.Z3)

        print("\nPrediction Probabilities")
        print(self.y_hat)

        return self.y_hat

    # RELU INVESTIGATION
    def relu_report(self):

        print("\n" + "=" * 60)
        print("RELU INVESTIGATION REPORT")
        print("=" * 60)

        print("\nHidden Layer 1")

        for i, value in enumerate(self.Z1[0]):

            if value > 0:

                print(
                    f"Neuron {i+1} : ACTIVE -> Gradient Allowed"
                )

            else:

                print(
                    f"Neuron {i+1} : DEAD -> Gradient Blocked"
                )

        print("\nHidden Layer 2")

        for i, value in enumerate(self.Z2[0]):

            if value > 0:

                print(
                    f"Neuron {i+1} : ACTIVE -> Gradient Allowed"
                )

            else:

                print(
                    f"Neuron {i+1} : DEAD -> Gradient Blocked"
                )

    # GRADIENT ANALYSIS
    def analyze_gradient(
            self,
            gradient_matrix,
            layer_name):

        print("\n" + "=" * 60)
        print(f"GRADIENT ANALYSIS : {layer_name}")
        print("=" * 60)

        rows, cols = gradient_matrix.shape

        for i in range(rows):

            for j in range(cols):

                g = gradient_matrix[i][j]

                print(
                    f"\nConnection [{i},{j}]"
                )

                print(
                    f"Gradient = {g:.6f}"
                )

                if abs(g) < 0.001:

                    print(
                        "Contribution : Negligible"
                    )

                elif g > 0:

                    print(
                        "Weight Too Large"
                    )

                    print(
                        "Action : DECREASE"
                    )

                else:

                    print(
                        "Weight Too Small"
                    )

                    print(
                        "Action : INCREASE"
                    )

    # FEATURE IMPORTANCE
    def feature_importance(self, dW1):

        print("\n" + "=" * 60)
        print("FEATURE RESPONSIBILITY REPORT")
        print("=" * 60)

        feature_names = [
            "Technical Skill",
            "Communication",
            "Experience"
        ]

        importance = np.sum(
            np.abs(dW1),
            axis=1
        )

        for feature, score in zip(
                feature_names,
                importance):

            print(
                f"{feature:<20}: {score:.6f}"
            )


    # BACKPROPAGATION
    def backward(self, y_true):

        print("\n" + "=" * 60)
        print("BACKPROPAGATION INVESTIGATION")
        print("=" * 60)

        # OUTPUT ERROR (dZ3 = y_hat - y_true)
        dZ3 = self.y_hat - y_true

        print("\nOUTPUT ERROR REPORT")
        print(dZ3)

        # W3 GRADIENT
        dW3 = np.dot(
            self.A2.T,
            dZ3
        )

        db3 = np.sum(
            dZ3,
            axis=0,
            keepdims=True
        )

        # ERROR FLOW TO HIDDEN LAYER 2
        dA2 = np.dot(
            dZ3,
            self.W3.T
        )

        dZ2 = dA2 * self.relu_derivative(
            self.Z2
        )

        dW2 = np.dot(
            self.A1.T,
            dZ2
        )

        db2 = np.sum(
            dZ2,
            axis=0,
            keepdims=True
        )

        # ERROR FLOW TO HIDDEN LAYER 1
        dA1 = np.dot(
            dZ2,
            self.W2.T
        )

        dZ1 = dA1 * self.relu_derivative(
            self.Z1
        )

        dW1 = np.dot(
            self.X.T,
            dZ1
        )

        db1 = np.sum(
            dZ1,
            axis=0,
            keepdims=True
        )

        # REPORTS
        self.relu_report()

        self.analyze_gradient(
            dW3,
            "OUTPUT LAYER (W3)"
        )

        self.analyze_gradient(
            dW2,
            "HIDDEN LAYER 2 (W2)"
        )

        self.analyze_gradient(
            dW1,
            "HIDDEN LAYER 1 (W1)"
        )

        self.feature_importance(dW1)

        # WEIGHT UPDATE DEMO
        learning_rate = 0.01

        print("\n" + "=" * 60)
        print("WEIGHT UPDATE SIMULATION")
        print("=" * 60)

        old_weight = self.W3[0][0]

        gradient = dW3[0][0]

        new_weight = (
            old_weight
            -
            learning_rate * gradient
        )

        print(
            f"\nOld Weight : {old_weight:.6f}"
        )

        print(
            f"Gradient   : {gradient:.6f}"
        )

        print(
            f"New Weight : {new_weight:.6f}"
        )

        print(
            f"Change     : {(new_weight-old_weight):.6f}"
        )

        # FINAL SUMMARY
        print("\n" + "=" * 60)
        print("FINAL INVESTIGATION SUMMARY")
        print("=" * 60)

        print("""
1. Prediction generated.

2. Error calculated using:
   dZ3 = y_hat - y_true

3. Error propagated backwards.

4. ReLU blocked inactive neurons.

5. dW3 identified final decision errors.

6. dW2 identified concept-level errors.

7. dW1 identified feature-level errors.

8. Positive Gradient
      -> Reduce Weight

9. Negative Gradient
      -> Increase Weight

10. Gradient Descent will use
    these gradients to learn.
        """)

        return dW1, dW2, dW3
```

```python
# MAIN PROGRAM
X = np.array([[2, 3, 1]])

# Actual Class = Hire
# One Hot Encoding

y_true = np.array([[0, 1]])

model = NeuralNetworkInvestigation()

prediction = model.forward(X)

model.backward(y_true)
```
