## What this skill is about
Retrieve the most relevant stored embeddings by maximum inner-product search with plain NumPy.

## Objectives
- Represent documents as embedding vectors in one shared dimension.
- Compute query-to-document inner products.
- Rank documents and return the top-k results.

Run the cells top to bottom. Requires `torch`, `numpy` and `matplotlib` (`pip install torch numpy matplotlib`).

## Theory and core formulas
$$s=Dq,\qquad \text{top-}k=\text{argsort}_{\downarrow}(s)[:k]$$
$q\cdot d=\lVert q\rVert\lVert d\rVert\cos\theta$, so MIPS differs from cosine unless vectors are normalised.

## Implementation
Reference code from the lab manual (unchanged).

```python
import numpy as np

def mips_search(query_embedding, document_embeddings, documents, top_k=3):
    q = np.asarray(query_embedding, dtype=np.float32)
    D = np.asarray(document_embeddings, dtype=np.float32)
    if D.ndim != 2 or q.ndim != 1 or D.shape[1] != q.shape[0]:
        raise ValueError("Embedding dimensions do not match")
    scores = D @ q
    top_indices = np.argsort(scores)[::-1][:top_k]
    return [(documents[i], float(scores[i])) for i in top_indices]

# Synthetic embeddings for a runnable smoke test.
rng = np.random.default_rng(7)
docs = [
    "Attention relates queries to keys and values.",
    "Diffusion gradually adds noise to data.",
    "LoRA adapts a frozen weight matrix efficiently.",
    "Causal masks block future-token attention."
]
doc_vectors = rng.normal(size=(len(docs), 8)).astype(np.float32)
query = rng.normal(size=8).astype(np.float32)
for doc, score in mips_search(query, doc_vectors, docs, top_k=2):
    print(f"{score:.4f}  {doc}")
```

## What to check
**Expected result:** Top-k documents returned in descending raw dot-product score (random embeddings, seeded).

## Pitfalls
Use the same embedding model for query and documents. Normalise if you want cosine behaviour. Rebuild the index when the model or chunking changes.
