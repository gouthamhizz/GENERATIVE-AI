# Skill 33: Sinusoidal Positional Encoding

**Phase 3 - LLMs & Generative AI Applications (Skills 31-45)**

## What this skill is about
Generate absolute positional encodings from sines and cosines and explain why they work.

## Objectives
- Explain why self-attention alone does not encode token order.
- Implement alternating sine and cosine channels.
- Visualize how positions and embedding dimensions vary.

Run the cells top to bottom. Requires `torch`, `numpy` and `matplotlib` (`pip install torch numpy matplotlib`).

## Theory and core formulas
$$PE(p,2i)=\sin\!\big(p/10000^{2i/E}\big),\qquad PE(p,2i+1)=\cos\!\big(p/10000^{2i/E}\big)$$
$PE(p+k)$ is a fixed rotation of $PE(p)$, so relative offsets are linearly accessible.

## Implementation
Reference code from the lab manual (unchanged).

```python
import torch
import matplotlib.pyplot as plt

def sinusoidal_positional_encoding(max_len, d_model):
    pe = torch.zeros(max_len, d_model)
    positions = torch.arange(max_len, dtype=torch.float32).unsqueeze(1)
    div_term = torch.exp(
        torch.arange(0, d_model, 2, dtype=torch.float32)
        * (-torch.log(torch.tensor(10000.0)) / d_model)
    )
    pe[:, 0::2] = torch.sin(positions * div_term)
    # Handles odd d_model by matching the number of odd columns.
    pe[:, 1::2] = torch.cos(positions * div_term[:pe[:, 1::2].shape[1]])
    return pe

pe = sinusoidal_positional_encoding(80, 32)
plt.figure(figsize=(10, 5))
plt.imshow(pe.numpy(), aspect="auto", cmap="coolwarm")
plt.xlabel("Embedding dimension")
plt.ylabel("Token position")
plt.title("Sinusoidal positional encoding")
plt.colorbar()
plt.tight_layout()
plt.show()
print(pe.shape)  # (80, 32)
```

## What to check
**Expected result:** Matrix shape `(max_len, d_model)`; the heatmap shows different frequencies across columns.

## Pitfalls
Odd `d_model` needs the sliced `div_term`. Position vectors must be on a scale comparable to the embeddings. Indexing beyond `max_len` fails.
