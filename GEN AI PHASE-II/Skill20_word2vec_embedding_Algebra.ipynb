#import required library


```python
import numpy as np
```

# Load GloVe Embeddings

```python
embedding_file = "/content/glove.6B.100d.txt"

embeddings = {}

with open(embedding_file, "r", encoding="utf8") as f:

    for line in f:

        values = line.split()

        word = values[0]

        vector = np.asarray(values[1:], dtype=np.float32)

        embeddings[word] = vector

print("Vocabulary Size:", len(embeddings))
```

# Cosine Similarity

```python
def cosine_similarity(vec1, vec2):

    return np.dot(vec1, vec2) / (
        np.linalg.norm(vec1)
        * np.linalg.norm(vec2)
    )
```

# Find Most Similar Word

```python
def most_similar(vector, exclude=[]):

    best_word = None
    best_score = -1

    for word, vec in embeddings.items():

        if word in exclude:
            continue

        score = cosine_similarity(vector, vec)

        if score > best_score:

            best_score = score
            best_word = word

    return best_word, best_score
```

# Embedding Algebra


```python
result_vector = (
    embeddings["king"]
    - embeddings["man"]
    + embeddings["woman"]
)

word, score = most_similar(
    result_vector,
    exclude=["king", "man", "woman"]
)

print("\nAnalogy Result")
print("------------------")
print("king - man + woman =", word)
print("Cosine Similarity:", round(score, 4))
```


# Similarity Examples


```python
pairs = [
    ("king", "queen"),
    ("man", "woman"),
    ("paris", "france"),
    ("rome", "italy")
]

print("\nCosine Similarities")

for w1, w2 in pairs:

    sim = cosine_similarity(
        embeddings[w1],
        embeddings[w2]
    )

    print(f"{w1:10s} {w2:10s} : {sim:.4f}")
```
