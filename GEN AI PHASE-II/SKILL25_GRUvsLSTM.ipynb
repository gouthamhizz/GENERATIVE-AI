#**GRU VS LSTM ABLATION**

BUILD A GATED RECURRENT UNIT (GRU) FROM SCRATCH. CONDUCT AN ABLATION STUDY COMPARING THE PARAMETER COUNT AND TRAINING VELOCITY AGAINST THE LSTM



```python
import numpy as np
import time
```

```python
# Activation Functions

def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
# Synthetic Input Sequence

X = np.array([
    [1.0, 0.5, 0.2],
    [0.3, 0.8, 0.1],
    [0.6, 0.4, 0.9],
    [0.2, 0.1, 0.5],
    [0.7, 0.3, 0.6]
])
```

```python
input_size = 3
hidden_size = 4

np.random.seed(42)

concat_size = input_size + hidden_size
```

```python
# GRU Parameters

Wz = np.random.randn(hidden_size, concat_size)
Wr = np.random.randn(hidden_size, concat_size)
Wh = np.random.randn(hidden_size, concat_size)

bz = np.random.randn(hidden_size,1)
br = np.random.randn(hidden_size,1)
bh = np.random.randn(hidden_size,1)
```

```python
# Initial Hidden State

h = np.zeros((hidden_size,1))
```

```python
# Forward Pass

start = time.perf_counter()

for t in range(len(X)):

    x = X[t].reshape(input_size,1)

    concat = np.vstack((h, x))

    # Update Gate
    z = sigmoid(Wz @ concat + bz)

    # Reset Gate
    r = sigmoid(Wr @ concat + br)

    # Candidate Hidden State
    reset_hidden = r * h

    candidate_input = np.vstack((reset_hidden, x))

    h_tilde = np.tanh(
        Wh @ candidate_input + bh
    )

    # Hidden State Update
    h = (1-z) * h + z * h_tilde

    print(f"\nTime Step {t+1}")

    print("Update Gate")
    print(z)

    print("Reset Gate")
    print(r)

    print("Candidate Hidden")
    print(h_tilde)

    print("Hidden State")
    print(h)

end = time.perf_counter()

print("\nExecution Time")

print(f"{end-start:.6f} seconds")
```

```python
# Parameter Count
gru_params = (
    Wz.size + Wr.size + Wh.size +
    bz.size + br.size + bh.size
)

lstm_params = (
    4*(hidden_size*concat_size)
    +4*hidden_size
)

print("\nGRU Parameters :", gru_params)

print("Equivalent LSTM Parameters :", lstm_params)
```
