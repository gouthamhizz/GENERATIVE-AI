**LSTM CELL MATHEMATICS**

IMPLEMENT AN LSTM CELL COMPLETELY FROM SCRATCH IN PYTHON. EXPLICITLY CODE THE INPUT, FORGET, OUTPUT GATES, AND THE CELL STATE UPDATE EQUATIONS.

```python
import numpy as np
```

```python
# Activation Functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
# Synthetic Input Sequence

X = np.array([
    [1.0, 0.5, 0.2],
    [0.3, 0.8, 0.1],
    [0.6, 0.4, 0.9],
    [0.2, 0.1, 0.5],
    [0.7, 0.3, 0.6]
])

input_size = 3
hidden_size = 4

np.random.seed(42)
```

```python
# LSTM Parameters

concat_size = input_size + hidden_size

Wf = np.random.randn(hidden_size, concat_size)
Wi = np.random.randn(hidden_size, concat_size)
Wc = np.random.randn(hidden_size, concat_size)
Wo = np.random.randn(hidden_size, concat_size)

bf = np.random.randn(hidden_size,1)
bi = np.random.randn(hidden_size,1)
bc = np.random.randn(hidden_size,1)
bo = np.random.randn(hidden_size,1)
```

```python
# Initial States

h = np.zeros((hidden_size,1))
C = np.zeros((hidden_size,1))
```

```python
# Forward Pass

for t in range(len(X)):

    x = X[t].reshape(input_size,1)

    concat = np.vstack((h, x))

    # Forget Gate
    f = sigmoid(Wf @ concat + bf)

    # Input Gate
    i = sigmoid(Wi @ concat + bi)

    # Candidate Cell
    C_tilde = np.tanh(Wc @ concat + bc)

    # Cell State Update
    C = f * C + i * C_tilde

    # Output Gate
    o = sigmoid(Wo @ concat + bo)

    # Hidden State
    h = o * np.tanh(C)

    print(f"\nTime Step {t+1}")

    print("Forget Gate")
    print(f)

    print("Input Gate")
    print(i)

    print("Cell State")
    print(C)

    print("Output Gate")
    print(o)

    print("Hidden State")
    print(h)

print("\nFinal Hidden State")
print(h)

print("\nFinal Cell State")
print(C)
```
