#SKILL 23 : GRADIENT CLIPPING

IMPLEMENT L2-NORM GRADIENT CLIPPING TO MATHEMATICALLY PREVENT THE EXPLOADING GRADIENT PHENOMENON DURING BPTT




```python
import numpy as np
```

# Example Gradients from BPTT

```python
np.random.seed(42)

dWxh = np.random.randn(4, 3) * 15
dWhh = np.random.randn(4, 4) * 15
dBh = np.random.randn(4, 1) * 15
```

# Compute Global L2 Norm

```python
def compute_global_norm(gradients):

    total = 0.0

    for grad in gradients:
        total += np.sum(grad ** 2)

    return np.sqrt(total)
```

# L2 Gradient Clipping

```python
def clip_gradients(gradients, threshold):

    global_norm = compute_global_norm(gradients)

    print(f"Gradient Norm Before Clipping : {global_norm:.4f}")

    if global_norm > threshold:

        scale = threshold / global_norm

        print(f"Scaling Factor : {scale:.4f}")

        for grad in gradients:
            grad *= scale

    global_norm_after = compute_global_norm(gradients)

    print(f"Gradient Norm After Clipping  : {global_norm_after:.4f}")

    return gradients
```

# Apply Gradient Clipping And Displaying Result

```python
# Apply Gradient Clipping

threshold = 5.0

gradients = [dWxh, dWhh, dBh]

clip_gradients(gradients, threshold)

# Display Results

print("\ndWxh\n")
print(dWxh)

print("\ndWhh\n")
print(dWhh)

print("\ndBh\n")
print(dBh)
```
