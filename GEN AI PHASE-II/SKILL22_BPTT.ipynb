#SKILL 22: BACKPROPAGATION THROUGH TIME (BPTT)

# DERIVE AND CODE THE UNROLLED GRADIENT ACCUMULATION ACROSS 5 TIME STEPS FOR THE RNN CELL FOR SKILL 21

```python
import numpy as np
```

# Parameters

```python
np.random.seed(42)

T = 5
input_size = 3
hidden_size = 4

X = np.array([
    [1.0,0.5,0.2],
    [0.3,0.8,0.1],
    [0.6,0.4,0.9],
    [0.2,0.1,0.5],
    [0.7,0.3,0.6]
])

target = np.array([[0.5],[-0.2],[0.8],[0.1]])

Wxh = np.random.randn(hidden_size,input_size)

Whh = np.random.randn(hidden_size,hidden_size)

Bh = np.random.randn(hidden_size,1)
```

# Forward Pass

```python
h_states = [np.zeros((hidden_size,1))]

for t in range(T):

    x = X[t].reshape(input_size,1)

    h = np.tanh(
        Wxh @ x +
        Whh @ h_states[-1] +
        Bh
    )

    h_states.append(h)
```

# Loss

```python
loss = 0.5 * np.sum(
    (h_states[-1]-target)**2
)

print("Loss =", loss)
```

# Initialize Gradients

```python
dWxh = np.zeros_like(Wxh)
dWhh = np.zeros_like(Whh)
dBh = np.zeros_like(Bh)

dh_next = h_states[-1] - target
```

# BPTT

```python
for t in reversed(range(T)):

    h = h_states[t+1]
    h_prev = h_states[t]
    x = X[t].reshape(input_size,1)

    dtanh = (
        dh_next *
        (1 - h*h)
    )

    dWxh += dtanh @ x.T
    dWhh += dtanh @ h_prev.T
    dBh += dtanh

    dh_next = Whh.T @ dtanh
```

# Gradient Clipping

```python
for grad in [dWxh,dWhh,dBh]:

    np.clip(
        grad,
        -5,
        5,
        out=grad
    )
```

# Display

```python
print("\ndWxh\n")
print(dWxh)

print("\ndWhh\n")
print(dWhh)

print("\ndBh\n")
print(dBh)
```
