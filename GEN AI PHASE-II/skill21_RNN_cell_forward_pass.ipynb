# skill 21 : RNN Cell forward pass

Implement a vanilla recurrent neural network hidden state , update from scratch using the formula : Ht=tanh(Whh Ht-1 + Wxh Xt + Bh)

```python
import numpy as np
```

```python
X = np.array([
    [1.0, 0.5, 0.2],
    [0.3, 0.8, 0.1],
    [0.6, 0.4, 0.9],
    [0.2, 0.1, 0.5],
    [0.7, 0.3, 0.6]
])
```

```python
# Dimensions
input_size = 3
hidden_size = 4

np.random.seed(42)
```

```python
# Initialize Parameters

Wxh = np.random.randn(hidden_size, input_size)

Whh = np.random.randn(hidden_size, hidden_size)

Bh = np.random.randn(hidden_size, 1)
```

```python
# Initial Hidden State

h_prev = np.zeros((hidden_size, 1))
```

```python
# Forward Pass
hidden_states = []

for t in range(len(X)):

    x_t = X[t].reshape(input_size, 1)

    h_t = np.tanh(
        np.dot(Wxh, x_t) +
        np.dot(Whh, h_prev) +
        Bh
    )

    hidden_states.append(h_t)

    print(f"\nTime Step {t+1}")

    print("Input")

    print(x_t)

    print("Hidden State")

    print(h_t)

    h_prev = h_t

print("\nFinal Hidden State")

print(hidden_states[-1])
```
